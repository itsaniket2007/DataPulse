### Visualization Fundamentals

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/cleaned/sales_data_cleaned.csv")

df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
# Visualization 1) Product Performance
product_sales = (
    df.groupby("Product")["Net_Sales"]
      .sum()
      .sort_values(ascending=False)
)
```

```python
plt.figure(figsize=(10, 5))

plt.bar(product_sales.index, product_sales.values)

plt.title("Net Sales by Product")
plt.xlabel("Product")
plt.ylabel("Net Sales (₹)")

plt.xticks(rotation=30)

plt.show()
```

Net Sales by Product — 

The chart compares the total net sales generated by each product.

Key observation:

Laptop is the clear top-performing product, generating approximately ₹127,000 in net sales.
Monitor ranks second with around ₹55,000.
Headphones generate approximately ₹13,000, while Mouse and Keyboard contribute relatively little.
Laptop sales are more than 2× the sales of Monitors and far exceed the remaining products.

Business implication:

Laptops are the primary revenue driver and should receive strong inventory and sales attention.
The large gap between laptops and other products suggests the business is heavily dependent on laptop revenue.
The business could investigate whether monitors and accessories can be bundled with laptop purchases to increase their sales and overall revenue.
Low-performing products such as Keyboard and Mouse should be analyzed further to determine whether pricing, demand, inventory, or product positioning is responsible.

```python
# Visualization  2 — Category Performance
category_sales = (
    df.groupby("Category")["Net_Sales"]
      .sum()
      .sort_values(ascending=False)
)
```

```python
plt.figure(figsize=(7, 5))

plt.bar(category_sales.index, category_sales.values)

plt.title("Net Sales by Category")
plt.xlabel("Category")
plt.ylabel("Net Sales (₹)")

plt.show()
```

```python
# Visualization  3 — Regional Performance
region_sales = (
    df.groupby("Region")["Net_Sales"]
      .sum()
      .sort_values(ascending=False)
)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(region_sales.index, region_sales.values)

plt.title("Net Sales by Region")
plt.xlabel("Region")
plt.ylabel("Net Sales (₹)")

plt.show()
```

```python
# Visualization #4 — Revenue Over Time
daily_sales = (
    df.groupby("Order_Date")["Net_Sales"]
      .sum()
      .sort_index()
)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(daily_sales.index, daily_sales.values)

plt.title("Daily Net Sales")
plt.xlabel("Order Date")
plt.ylabel("Net Sales (₹)")

plt.xticks(rotation=45)

plt.show()
```

```python
# grid to the line chart
plt.figure(figsize=(12, 5))

plt.plot(daily_sales.index, daily_sales.values)

plt.title("Daily Net Sales")
plt.xlabel("Order Date")
plt.ylabel("Net Sales (₹)")

plt.xticks(rotation=45)
plt.grid()

plt.show()
```

Daily Net Sales — 

The chart shows the daily net sales trend over the observed period.

Key observation:

Net sales fluctuate significantly from day to day, with no consistent upward or downward trend.
The highest sales occur around January 18, reaching approximately ₹65,000.
Another major peak appears at the beginning of the period, at around ₹62,000.
Several days have very low sales, around ₹2,000–₹5,000.
There are noticeable sales spikes around January 6, January 13, and January 18.

Business implication:

The large fluctuations suggest that sales may be influenced by specific high-performing days, promotions, customer demand, or other external factors.
The business should investigate what happened on the peak-sales dates to identify repeatable factors driving high revenue.
Comparing these peaks with product, category, payment method, or order volume could help explain the variation and improve sales forecasting.

```python
# Payment Count Visulaization 
payment_counts = df["Payment_Method"].value_counts()
plt.figure(figsize=(12, 5))

plt.bar(payment_counts.index, payment_counts.values)

plt.title("Overall Payment Count")
plt.xlabel("Payment Method")
plt.ylabel("Number of Transactions")
plt.show()
```

The chart shows the overall number of transactions made through each payment method.

Key observation:

UPI is the most frequently used payment method with 7 transactions.
Credit Card, Debit Card, and Cash each account for 2 transactions.
UPI usage is significantly higher than every other payment method.

Business implication:

Customers show a strong preference for UPI, suggesting that digital payments are highly popular.
The business should ensure that UPI payment infrastructure remains reliable and easily accessible.
Since card and cash usage is comparatively low, future analysis could investigate why customers prefer UPI and whether this pattern varies by product, location, or customer segment.
