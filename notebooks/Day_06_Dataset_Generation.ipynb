```python
import pandas as pd
import numpy as np
import random
```

```python
# business dimensions
products = {
    "Laptop": ("Electronics", 65000),
    "Monitor": ("Electronics", 15000),
    "Keyboard": ("Accessories", 2000),
    "Mouse": ("Accessories", 800),
    "Headphones": ("Accessories", 2500),
    "Webcam": ("Accessories", 3500),
    "Tablet": ("Electronics", 30000),
    "Smartphone": ("Electronics", 40000)
}
```

```python
# Creating Coustmer , We're going to generate 500 customers.
customer_ids = [
    f"C{str(i).zfill(4)}"
    for i in range(1, 501)
]
```

```python
# Create regions
regions = [
    "North",
    "South",
    "East",
    "West"
]
# Payment methods:
payment_methods = [
    "UPI",
    "Credit Card",
    "Debit Card",
    "Cash"
]
```

```python
# Generate orders , We're going to create 5,000 orders.
num_orders = 5000
order_ids = [
    f"O{str(i).zfill(5)}"
    for i in range(1, num_orders + 1)
]
```

```python
# Generating Dates 
dates = pd.date_range(
    start="2025-01-01",
    end="2026-06-30",
    periods=num_orders
)
```

```python
# generate the orders 
data = []

for order_id, order_date in zip(order_ids, dates):

    customer_id = random.choice(customer_ids)

    product = random.choice(list(products.keys()))

    category, unit_price = products[product]

    region = random.choice(regions)

    quantity = random.randint(1, 5)

    discount = random.choice([
        0.00,
        0.05,
        0.10,
        0.15
    ])

    payment_method = random.choice(payment_methods)

    returned = random.choices(
        ["Yes", "No"],
        weights=[8, 92]
    )[0]

    data.append([
        order_id,
        order_date,
        customer_id,
        product,
        category,
        region,
        quantity,
        unit_price,
        discount,
        payment_method,
        returned
    ])
```

```python
# Convert into DataFrame
columns = [
    "Order_ID",
    "Order_Date",
    "Customer_ID",
    "Product",
    "Category",
    "Region",
    "Quantity",
    "Unit_Price",
    "Discount",
    "Payment_Method",
    "Returned"
]
large_df = pd.DataFrame(
    data,
    columns=columns
)
large_df.head()
```

```python
# Adding Sales Metrics 
large_df["Gross_Sales"] = (
    large_df["Quantity"] *
    large_df["Unit_Price"]
)
large_df["Discount_Amount"] = (
    large_df["Gross_Sales"] *
    large_df["Discount"]
)
large_df["Net_Sales"] = (
    large_df["Gross_Sales"] -
    large_df["Discount_Amount"]
)
```
